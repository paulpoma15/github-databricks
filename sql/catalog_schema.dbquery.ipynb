```sql
-- 1. Total de atletas y eventos registrados por Federación (Agrupación)
SELECT 
    FEDERACION,
    COUNT(DISTINCT DEPORTISTA) AS total_deportistas,
    COUNT(DISTINCT EVENTO) AS total_eventos
FROM big_data_team3.ciro.deportistas_gold
GROUP BY FEDERACION
ORDER BY total_deportistas DESC;

-- 2. Top 10 países con mayor cantidad de eventos organizados (Ranking y Filtros)
SELECT 
    PAIS_EVENTO,
    COUNT(DISTINCT EVENTO) AS total_eventos
FROM big_data_team3.ciro.deportistas_gold
WHERE PAIS_EVENTO IS NOT NULL
GROUP BY PAIS_EVENTO
ORDER BY total_eventos DESC
LIMIT 10;

-- 3. Distribución de participaciones (Comparativas: Individual vs Colectivo)
SELECT 
    COLECTIVO,
    COUNT(1) AS total_participaciones,
    ROUND(COUNT(1) * 100.0 / SUM(COUNT(1)) OVER(), 2) AS porcentaje
FROM big_data_team3.ciro.deportistas_gold
WHERE COLECTIVO IS NOT NULL
GROUP BY COLECTIVO;

-- 4. Deportistas con mayor cantidad de primeros lugares / Puesto 1 / ORO (Ranking)
SELECT 
    DEPORTISTA,
    FEDERACION,
    COUNT(1) AS primeros_puestos
FROM big_data_team3.ciro.deportistas_gold
WHERE TRY_CAST(PUESTO AS INT) = 1 OR UPPER(PUESTO) = 'ORO'
GROUP BY DEPORTISTA, FEDERACION
ORDER BY primeros_puestos DESC
LIMIT 10;

-- 5. Cantidad de deportistas agrupados por Departamento de origen (Agrupación)
SELECT 
    DEPARTAMENTO,
    COUNT(1) AS total_deportistas
FROM big_data_team3.ciro.deportistas_gold
WHERE DEPARTAMENTO IS NOT NULL
GROUP BY DEPARTAMENTO
ORDER BY total_deportistas DESC;
```
