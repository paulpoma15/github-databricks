```python
# 1. Leer el CSV usando el protocolo s3:// mediante Unity Catalog
df = spark.read.option("header", "true") \
               .option("sep", ";") \
               .option("inferSchema", "true") \
               .csv("s3://ciro-poma-databricks-2026/Data_DeportistasEventos_0.csv")

# 2. Registrar la tabla Delta en Unity Catalog
df.write.mode("overwrite").saveAsTable("big_data_team3.ciro.deportistas_gold")

print("¡Éxito! Ingesta nativa completada en big_data_team3.ciro.deportistas_gold")
```
